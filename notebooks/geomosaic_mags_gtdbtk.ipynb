## Downstream analysis for MAGs-gtdbtk 
#### module: mags_tax_annotation

GTDBT-k annotations

### Set-up the environemnt and metadata files

Create and activate the conda environment
```bash
conda env create -f environment_gtdbtk_tree.yaml

conda activate tree_utils
```

Download the R220 Bacteria Metadata (approx 184MB)
```bash
wget https://data.gtdb.ecogenomic.org/releases/release220/220.0/bac120_metadata_r220.tsv.gz
```
Download the R220 Archaea Metadata (approx 4M)
```bash
wget https://data.gtdb.ecogenomic.org/releases/release220/220.0/ar53_metadata_r220.tsv.gz
```

### Collect and prepare geomosaic's outputs:
- gtdbtk.bac120.classify.tree*
- gtdbtk.bac120.summary.tsv
- gtdbtk.ar53.summary.tsv
- gtdbtk.ar53.classify.tree*

 ```bash
 cp gtdbtk.bac120.classify.tree* trees_bac120/
 cp gtdbtk.ar53.classify.tree* trees_ar53/
 ```

Execute the command below to get the geomosaic MAGs ids:
```bash
awk -F'\t' 'NR>1 {print $1"\t"$1}' gtdbtk.bac120.summary.tsv > id_to_name_bac120.map
```

```bash
awk -F'\t' 'NR>1 {print $1"\t"$1}' gtdbtk.ar53.summary.tsv > id_to_name_ar53.map
```

### Tree trimming and file preparation for iTOL visualization


Trimming tree

```bash 
python build_and_trim_tree.py /
    --tree_dir samples/CR18_ER180415_F/mags_gtdbtk/classify/ /
    --id_map samples/CR18_ER180415_F/mags_gtdbtk/id_to_name_bac120.map /
    --summary samples/CR18_ER180415_F/mags_gtdbtk/gtdbtk.bac120.summary.tsv /
    --trim_script trim_tree_to_target_leaves.py /
    --archaea_meta ar53_metadata_r220.tsv /
    --bacteria_meta bac120_metadata_r220.tsv /
    --outdir  samples/CR18_ER180415_F/trimmed_trees
```

Generating iTOL files for taxa annotation and colouring

```bash
python3 make_itol_datasets.py /
    -s samples/CR18_ER180415_F/mags_gtdbtk/gtdbtk.bac120.summary.tsv /
    -m bac120_metadata_r220.tsv
```
