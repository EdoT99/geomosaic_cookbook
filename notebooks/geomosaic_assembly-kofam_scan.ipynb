## Downstream analysis for Assembly-KOFAM-scan with KEGG-DEcoder ##
```bash
geomosaic gather -s gmsetup.yaml -p kofam_scan
```
Create and activate the [KEGG-decoder](https://github.com/bjtully/BioData/blob/master/KEGGDecoder/README.md) environment available in the geomosaic_cookbook folder. A more exhaustive guide is available on their site.

```bash
conda env create -f kegg_decoder_environment.yaml
conda activate keggdecoder
```

Preprocessing geomosaic output file to accepted formatting for KEGG-decoder, since it does not accept multiple underscores in sample names
```bash
tail -n+2 gm_gathering/kofam_scan/geomosaic-kofam_scan-by-sample-long.csv |  awk -F "," '{print $2" "$1}' | sort -k1 | tr '_' '-' > kofam_input_kegg_decoder.txt

```


Execute the following line:
```bash
KEGG-decoder -i kofam_input_kegg_decoder.txt -o kegg_decoder_output.list -v static
```
